# Dataset Exploration — TweetEval Sentiment

This notebook analyses the raw social media dataset produced by `src/data_collection/download_dataset.py`.

````python
import json
import sys
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.config import RAW_CSV_PATH, EXPLORATION_RESULTS_JSON
from src.data_collection.explore_dataset import explore_dataset, print_summary, save_exploration_results

print("Project root:", PROJECT_ROOT)
print("Raw CSV:", RAW_CSV_PATH)
````

````python
df = pd.read_csv(RAW_CSV_PATH)
print(f"Rows: {len(df):,}  Columns: {list(df.columns)}")
df.head(10)
````

````python
stats = explore_dataset(df)
print_summary(stats)
save_exploration_results(stats)
stats
````

````python
print("Missing values per column:")
display(df.isna().sum().to_frame("missing"))

print("Duplicate rows:", int(df.duplicated().sum()))
print("Duplicate texts:", int(df["text"].duplicated().sum()))
````

````python
lengths = df["text"].fillna("").astype(str).str.len()
print(f"Average text length: {lengths.mean():.2f}")
print(f"Minimum text length: {lengths.min()}")
print(f"Maximum text length: {lengths.max()}")
lengths.describe()
````

````python
print("Sentiment distribution")
display(df["actual_sentiment"].value_counts().to_frame("count"))

print("Split distribution")
display(df["split"].value_counts().to_frame("count"))
````

## Notes

- `actual_sentiment` is the original TweetEval label (Negative / Neutral / Positive).
- It is kept for **evaluation only** and is not used as an input to VADER prediction.
- Exploration results are also written to `results/exploration_results.json`.
